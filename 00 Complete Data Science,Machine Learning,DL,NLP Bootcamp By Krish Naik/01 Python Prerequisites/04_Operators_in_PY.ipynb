```python
# Definition

# 👉 An operator is a special symbol that tells Python to perform a specific operation on values or variables.

# Types of Operators in Python

# 1. Arithmetic Operators
# 2. Comparison Operators
# 3. Logical Operators
# 4. Bitwise Operators
# 5. Assignment Operators
# 6. Identity Operators
# 7. Membership Operators
# 8. Special Operators
```

```python
# Arithmetic Operators 

# 👉 Arithmetic operators are used to perform mathematical operations.

# Here are the commonly used arithmetic operators in Python:

# 1. Addition (+)
# 2. Subtraction (-)
# 3. Multiplication (*)
# 4. Division (/)
# 5. Modulus (%)
# 6. Exponentiation (**)
# 7. Floor Division (//)
```

```python
a = 10
b = 15

print("Addition:", a + b)          # Output: 25
print("Subtraction:", a - b)       # Output: -5
print("Multiplication:", a * b)    # Output: 150
print("Division:", a / b)         # Output: 0.6666666666666666
print("Modulus:", a % b)          # Output: 10
print("Exponentiation:", a ** b)  # Output: 100
print("Floor Division:", a // b)  # Output: 0

```

```python
# Comparison Operators

# 👉 Comparison operators are used to compare two values.

# Here are the commonly used comparison operators in Python:

# 1. Equal to (==)
# 2. Not equal to (!=)
# 3. Greater than (>)
# 4. Less than (<)
# 5. Greater than or equal to (>=)
# 6. Less than or equal to (<=)
```

```python
a = 10
b = 10

print("Equal to:", a == b)               # Output: True
print("Not equal to:", a != b)          # Output: False
print("Greater than:", a > b)           # Output: False
print("Less than:", a < b)              # Output: False
print("Greater than or equal to:", a >= b)  # Output: True
print("Less than or equal to:", a <= b)     # Output: True
```

```python
# Logical Operators

# 👉 Logical operators are used to combine conditional statements.

# Here are the commonly used logical operators in Python:

# 1. Logical AND (and)
# 2. Logical OR (or)
# 3. Logical NOT (not)
```

```python
x = True
y = True

print("Logical AND:", x and y)  # Output: False
print("Logical OR:", x or y)    # Output: True
print("Logical NOT:", not x)     # Output: False
```

```python
fruits = ["apple", "banana"]

print("apple" in fruits)      # True
print("mango" not in fruits)  # True
print("banana" not in fruits)     # False
```
