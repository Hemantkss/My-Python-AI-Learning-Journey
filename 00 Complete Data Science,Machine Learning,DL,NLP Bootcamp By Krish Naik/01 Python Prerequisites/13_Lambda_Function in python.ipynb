```python
# A lambda function is a small anonymous (nameless) function in Python.
# It is used for short, one-line functions where using def would be unnecessary.

# Why use Lambda?

# Used when function is small & simple

# Mostly used with map(), filter(), reduce()

# Can be defined inside another function
```

```python
# Normal function
def add(a, b):
    return a + b

# Lambda function
add_lambda = lambda a, b: a + b

print(add(2, 3))         # 5
print(add_lambda(2, 3))  # 5
```

```python
#  (a) Single parameter
square = lambda x: x ** 2
print(square(5))   # 25
```

```python
# Multiple parameters
multiply = lambda a, b: a * b
print(multiply(3, 4))   # 12
```

```python
# No parameter
greet = lambda : "Hello Python"
print(greet())  # Hello Python
```

```python
# Using with map()

# Apply a function to each element of a list. 

nums = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x**2, nums))
print(squares)  # [1, 4, 9, 16, 25]
```

```python
# Using with filter()

# Filter elements based on condition.
nums = [10, 15, 20, 25, 30]
even = list(filter(lambda x: x % 2 == 0, nums))
print(even)  # [10, 20, 30]
```
