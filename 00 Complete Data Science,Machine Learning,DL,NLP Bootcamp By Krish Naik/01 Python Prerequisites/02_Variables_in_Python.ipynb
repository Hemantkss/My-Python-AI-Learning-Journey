```python
# Declaring and assigning variables
age = 25
height = 5.9
name = "John"
is_student = True

# Printing variables
print("Age:", age)
print("Height:", height)
print("Name:", name)
print("Is Student:", is_student)

# 
```

```python
# Case sensitivity in variable names
first_name = "Alice"
First_Name = "Bob"
FIRST_NAME = "Charlie"

print(first_name)
print(First_Name)
print(FIRST_NAME)
```

```python
#  Variables types
age = 25
height = 5.9
name = "John"
is_student = True

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

```python
# Type Checking and conversion
age = 25
print(type(age))

# Type conversion
age = str(age)
print(type(age))
```

```python
name = 'Hemant'
print(type(name))

# Type conversion
name = int(name)  # This will raise an error
print(type(name))
```

```python
# Dynamic Typing
age = 25
print(type(age))

age = "Twenty Five"
print(type(age))
```

```python
age = input("Enter your age: ")
name = input("Enter your name: ")

print("Hello, my name is", name, "and I am", age, "years old.")
print("Data type of age:", type(age))
print("Data type of name:", type(name))
```

```python
# Calculator 
num1 = input("Enter first number: ")
num2 = input("Enter second number: ")

# Performing addition
result = int(num1) + int(num2)
print("The sum is:", result)

# Performing subtraction
result = int(num1) - int(num2)
print("The difference is:", result)

# Performing multiplication
result = int(num1) * int(num2)
print("The product is:", result)

# Performing division
result = int(num1) / int(num2)
print("The quotient is:", result)
```
