```python
# A set is an unordered collection of unique elements in Python.

# Defined using {} (curly braces) or the set() constructor.

# Duplicates are automatically removed.

# my_set = {1, 2, 3, 3, 4}
# print(my_set)  # {1, 2, 3, 4}  (duplicate 3 removed)

# . Key Properties

# Unordered → no index, cannot access by position.

# Unique → no duplicate values allowed.

# Mutable → we can add/remove elements.

# Iterable → we can loop through elements.
```

```python
# Creating Sets  
# Using curly braces
fruits = {"apple", "banana", "cherry"}
print(fruits)

# Using set() constructor
numbers = set([1, 2, 3, 4])
print(numbers)

# empty_set = set()   # ✅ correct
# empty_dict = {}     # ❌ creates dictionary
```

```python
my_set = set([1, 2, 2, 3, 4, 4, 5, 6, 6])
print(my_set)  # {1, 2, 3, 4, 5, 6}
```

```python
# Set Operations
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# Union (A ∪ B)
print(A | B)        # {1, 2, 3, 4, 5, 6}
print(A.union(B))

# Intersection (A ∩ B)
print(A & B)        # {3, 4}
print(A.intersection(B))

# Difference (A - B)
print(A - B)        # {1, 2}
print(A.difference(B))

# Symmetric Difference (A Δ B)
print(A ^ B)        # {1, 2, 5, 6}
print(A.symmetric_difference(B))
```

```python
# Set Methods 
my_set = {1, 2, 3}

# Add element
my_set.add(4)
print(my_set)  # {1, 2, 3, 4}

# Update with multiple elements
my_set.update([5, 6])
print(my_set)  # {1, 2, 3, 4, 5, 6}

# Remove element
my_set.remove(2)   # ❌ Error if element not found
print(my_set)

my_set.discard(10)  # ✅ No error if not found

# # Pop random element
remove_el = my_set.pop()
print(f"Removed: {remove_el}")
print(my_set)

# # Clear all elements
my_set.clear()
print(my_set)  # set()

```

```python
# Membership Test
fruits = {"apple", "banana", "cherry"}

print("apple" in fruits)   # True
print("mango" not in fruits)  # True
print('kiwi' in fruits)  # False
```

```python
# Looping through Sets
for item in {"a", "b", "c"}:
    print(item)
```

```python
# Frozen Sets
fset = frozenset([1, 2, 3, 3])
print(fset)  # frozenset({1, 2, 3})

# fset.add(4)  #❌ Error (cannot modify)
```

```python
nums = [1, 2, 2, 3, 4, 4, 5]
unique_nums = list(set(nums))
print(unique_nums)  # [1, 2, 3, 4, 5]
```
