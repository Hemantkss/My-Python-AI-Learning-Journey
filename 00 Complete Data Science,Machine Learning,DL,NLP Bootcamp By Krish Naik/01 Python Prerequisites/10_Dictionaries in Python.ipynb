```python
# A Dictionary in Python is used to store data in key-value pairs.

# Each key is unique, and it maps to a value.

# Properties of Dictionaries

# Keys must be unique.

# Keys must be immutable (string, number, tuple).

# Values can be any type (string, number, list, dict, etc.).

# Dictionary is unordered (before Python 3.7), but in Python 3.7+ insertion order is preserved.
```

```python
student = {
    "name": "Rahul",
    "age": 20,
    "grade": "A"
}
print(student)
```

```python
# Accessing Dictionary Elements
print(student["name"])   # Rahul
print(student.get("age")) # 20

print(student.get("address", "Not Found"))  # Not Found
```

```python
# Updating Dictionary
student["age"] = 21   # update value
student["city"] = "Mumbai"  # add new key-value pair
print(student)
```

```python
# Removing Elements

student.pop("grade")   # removes a key
print(student)

student.popitem()      # removes last inserted item
print(student)

del student["age"]     # deletes key
print(student)

student.clear()        # clears dictionary
print(student)         # {}
```

```python
# Looping Through Dictionary
student = {"name": "Hemant", "age": 25, "grade": "A"}

# Loop through keys
for key in student:
    print(key)

# Loop through values
for value in student.values():
    print(value)

# # Loop through key-value pairs
for key, value in student.items():
    print(key, ":", value)
```

```python
# Dictionary Methods
print(student.keys() )   # returns all keys
student.values()  # returns all values
student.items()   # returns all key-value pairs
student.update({"age": 22, "city": "Delhi"})  # update/add
print(student)
```

```python
# Nested Dictionaries
students = {
    "student1": {"name": "Rahul", "age": 20},
    "student2": {"name": "Anita", "age": 22}
}
print(students["student1"]["name"])  # Rahul
```

```python
# Dictionary Comprehension
squares = {x: x*x for x in range(1, 6)}
print(squares)
```
