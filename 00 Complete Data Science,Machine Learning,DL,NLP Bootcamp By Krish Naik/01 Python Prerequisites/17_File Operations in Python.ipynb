```python
# Python provides built-in functions for creating, reading, writing, updating, and deleting files.
# The main function used is open().
```

```python
# Read / Open File
with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
# Read file line by line
with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip()) # Remove leading/trailing whitespace
```

```python
# Writing to a File (Overwriting existing content)
with open('example.txt', 'w') as file:
    file.write("Hello, World! \n")
    file.write("This is a new line.")
```

```python
# Write to a file (without overwriting)
with open('example.txt', 'a') as file:
    file.write("Hemant Sabale \n")
    file.write("Sp.")
```

```python
# Writing a list of lines to a file
lines = ["Hello, World! \n", "This is a new line. \n", "Hemant Sabale \n", "Sp. \n"]
with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
# Binary File Operations
data = b'hemant sabale'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
# Read  binary file
with open('example.bin', 'rb') as file:
    data = file.read()
    print(data)
```

```python
# Read the content from source text file and write it to a destination text file
with open('example.txt', 'r') as source_file:
    content = source_file.read()
    
with open('destination.txt', 'w') as dest_file:
        dest_file.write(content)
```

```python
# Deleting file
import os 

if os.path.exists('destination.txt'):
    os.remove('destination.txt')
else:
    print("The file does not exist")
```

```python
# Create a new directory 
import os
new_directory = "Text"
os.makedirs(new_directory, exist_ok=True)
print(f"Directory '{new_directory}' created.")
```

```python
# Listing Files and Directory
items = os.listdir('.')
for item in items:
    print(item)
```

```python
# Joining paths
dir_name = "Text"
file_name = "example.txt"
full_path = os.path.join(dir_name, file_name)
print(f"Full path: {full_path}")
```
