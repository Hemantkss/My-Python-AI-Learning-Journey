```python
# What are Magic Methods?

# Also called dunder methods (double underscore: __method__).

# They start and end with __ (e.g., __init__, __str__, __len__).

# They let you customize the behavior of objects.

# Example: len(obj) calls obj.__len__() internally.
```

```python
# a) __init__ → Object Initialization
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

b = Book("1984", "George Orwell")
print(b.title)   # 1984
```

```python
# b) __str__ → String Representation 
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
    
    def __str__(self):
        return f"{self.title} by {self.author}"

b = Book("1984", "George Orwell")
print(b)   # Calls __str__ → "1984 by George Orwell"
```

```python
# c) __len__ → Length of Object 
class Library:
    def __init__(self, books):
        self.books = books
    
    def __len__(self):
        return len(self.books)

lib = Library(["1984", "Animal Farm", "Brave New World"])
print(len(lib))   # 3
```

```python
# d) __add__ → Operator Overloading (+) 
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __str__(self):
        return f"({self.x}, {self.y})"

v1 = Vector(2, 4)
v2 = Vector(5, -2)
print(v1 + v2)   # (7, 2)
```

```python
# e) __eq__ → Equality Check (==) 
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __eq__(self, other):
        return self.age == other.age

p1 = Person("Alice", 25)
p2 = Person("Bob", 25)

print(p1 == p2)  # True (compares age, not memory location)
```

```python
# Real-World Example → Shopping Cart 🛒  
class Item:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    
    def __str__(self):
        return f"{self.name}: ${self.price}"

    def __add__(self, other):
        return self.price + other.price


item1 = Item("Shoes", 100)
item2 = Item("Shirt", 50)

print(item1)       # Shoes: $100
print(item1 + item2)  # 150
```

```python
# 4. Some Useful Magic Methods List
# Method	Usage Example
# __init__	Object initialization
# __str__	print(obj) string
# __repr__	Debugging representation
# __len__	len(obj)
# __getitem__	Indexing (obj[i])
# __setitem__	Assigning (obj[i] = x)
# __delitem__	Deleting (del obj[i])
# __iter__	Loops (for x in obj)
# __next__	Iteration next value
# __eq__	== operator
# __lt__	< operator
# __add__	+ operator
# __sub__	- operator
# __mul__	* operator
# __call__	Make object callable like function
```

```python
# Python allows us to redefine how operators work for user-defined objects (classes).

# This is done using magic methods (dunder methods).

# Example: +, -, *, /, ==, <, etc. can be customized for objects.
```

```python
class Product:
    def __init__(self, product_id, name, price, quantity):
        self.product_id = product_id
        self.name = name
        self.price = price
        self.quantity = quantity

    # == (compare by product_id)
    def __eq__(self, other):
        return self.product_id == other.product_id

    # < (compare by price)
    def __lt__(self, other):
        return self.price < other.price

    # + (add stock)
    def __add__(self, other):
        if self.product_id == other.product_id:
            return Product(self.product_id, self.name,
                           self.price, self.quantity + other.quantity)
        else:
            raise ValueError("Cannot add different products!")

    # - (reduce stock)
    def __sub__(self, other):
        if self.product_id == other.product_id:
            return Product(self.product_id, self.name,
                           self.price, self.quantity - other.quantity)
        else:
            raise ValueError("Cannot subtract different products!")

    # * (total cost for given quantity)
    def __mul__(self, qty):
        return self.price * qty

    def __str__(self):
        return f"{self.name} (₹{self.price}, Qty: {self.quantity})"


# Example usage
p1 = Product(101, "Laptop", 50000, 10)
p2 = Product(101, "Laptop", 50000, 5)
p3 = Product(102, "Phone", 30000, 8)

# == operator
print("Same product?", p1 == p2)  # True (same ID)

# < operator
print("Is Phone cheaper than Laptop?", p3 < p1)  # True

# + operator (adding stock)
combined_stock = p1 + p2
print("Combined stock:", combined_stock)  # Laptop (₹50000, Qty: 15)

# - operator (selling items)
remaining_stock = p1 - p2
print("Remaining stock:", remaining_stock)  # Laptop (₹50000, Qty: 5)

# * operator (total cost)
print("Buying 3 laptops cost:", p1 * 3)  # 150000
```
