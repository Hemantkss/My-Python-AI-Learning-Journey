```python
# The map() function is used to apply a function to each element in an iterable (like list, tuple, etc.).
# It transforms the data without writing explicit loops.
```

```python
def square(x):
    return x ** 2

nums = [1, 2, 3, 4]
result = map(square, nums)

print(result)          # <map object at 0x...>
print(list(result))    # [1, 4, 9, 16]
```

```python
# Using map() with Lambda 
nums = [1, 2, 3, 4]
result = map(lambda x: x ** 2, nums)
print(list(result))  # [1, 4, 9, 16]
```

```python
# With Multiple Iterables

# If multiple iterables are given, map() applies the function to corresponding elements. 
a = [1, 2, 3]
b = [4, 5, 6]

result = map(lambda x, y: x + y, a, b)
print(list(result))  # [5, 7, 9]
```

```python
# Converting Data Types 
nums = ["1", "2", "3", "4"]
result = map(int, nums)  
print(list(result))  # [1, 2, 3, 4]
```

```python
# Nested Example 
nums = [1, 2, 3, 4, 5]
result = map(lambda x: "Even" if x % 2 == 0 else "Odd", nums)
print(list(result))  # ['Odd', 'Even', 'Odd', 'Even', 'Odd']
```

```python
names = ["alice", "bob", "charlie"]
result = map(lambda name: name.capitalize(), names)
print(list(result))  # ['Alice', 'Bob', 'Charlie']
```
