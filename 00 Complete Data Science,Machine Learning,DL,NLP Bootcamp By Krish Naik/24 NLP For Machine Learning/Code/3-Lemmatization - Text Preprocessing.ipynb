## Wordnet Lemmatizer
Lemmatization technique is like stemming. The output we will get after lemmatization is called ‘lemma’, which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example −


```python
# Q&A,chatbots,text summarization
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
'''
POS- Noun-n
verb-v
adjective-a
adverb-r
'''

lemmatizer.lemmatize("going", pos='v')
```

```python
words = ["eating", "eats", "eaten", "writing", "writes", "programming", "programs", "history", "finally", "finalized"]
```

```python
for word in words:
    print(word + " => " + lemmatizer.lemmatize(word))
```

```python
lemmatizer.lemmatize("goes",pos='v')
```

```python
lemmatizer.lemmatize("fairly", pos='v'), lemmatizer.lemmatize("sportingly")
```

```python
words = ["studies", "studying", "played", "better", "mice", "running"]

for word in words:
    print(f"{word} → {lemmatizer.lemmatize(word)}")
```

```python
print(lemmatizer.lemmatize("running", pos='v'))   # Verb
print(lemmatizer.lemmatize("better", pos='a'))    # Adjective
print(lemmatizer.lemmatize("studies", pos='v'))   # Verb
print(lemmatizer.lemmatize("played", pos='v'))    # Verb
```
