## About Dataset
Context
This is a small subset of dataset of Book reviews from Amazon Kindle Store category.

Content
5-core dataset of product reviews from Amazon Kindle Store category from May 1996 - July 2014. Contains total of 982619 entries. Each reviewer has at least 5 reviews and each product has at least 5 reviews in this dataset.
Columns

- asin - ID of the product, like B000FA64PK
- helpful - helpfulness rating of the review - example: 2/3.
- overall - rating of the product.
- reviewText - text of the review (heading).
- reviewTime - time of the review (raw).
- reviewerID - ID of the reviewer, like A3SPTOKDG7WBLN
- reviewerName - name of the reviewer.
- summary - summary of the review (description).
- unixReviewTime - unix timestamp.

Acknowledgements

## Preprocessing And Cleaning

```python
# positive review is 1 and negative review is 0

df['rating'] = df['rating'].apply(lambda x: 0 if x < 3 else 1)
```

```python
df['rating'].value_counts()
```

```python
# 1. Lower All the cases

df['reviewText'] = df['reviewText'].str.lower()
```

```python
df.head()
```

```python
# 📦 1. Import all required libraries
import re
import pandas as pd
from bs4 import BeautifulSoup
import nltk
from nltk.corpus import stopwords

# 📦 2. Download NLTK stopwords (only first time)
nltk.download('stopwords')

# 🧠 3. Make sure your DataFrame exists (df)
# Example:
# df = pd.read_csv("reviews.csv")   # <-- your dataset

# ⚙️ 4. Removing special characters
df['reviewText'] = df['reviewText'].apply(lambda x: re.sub('[^a-zA-Z0-9 ]+', '', str(x)))

# ⚙️ 5. Remove stopwords
stop_words = set(stopwords.words('english'))
df['reviewText'] = df['reviewText'].apply(
    lambda x: " ".join([word for word in x.split() if word.lower() not in stop_words])
)

# ⚙️ 6. Remove URLs
df['reviewText'] = df['reviewText'].apply(
    lambda x: re.sub(r'(http|https|ftp|ssh)://\S+', '', str(x))
)

# ⚙️ 7. Remove HTML tags
# Try lxml first, fallback to html.parser if not installed
try:
    df['reviewText'] = df['reviewText'].apply(lambda x: BeautifulSoup(x, 'lxml').get_text())
except Exception:
    df['reviewText'] = df['reviewText'].apply(lambda x: BeautifulSoup(x, 'html.parser').get_text())

# ⚙️ 8. Remove extra spaces
df['reviewText'] = df['reviewText'].apply(lambda x: " ".join(x.split()))

# ✅ 9. Preview cleaned text
print(df['reviewText'].head())

```

```python
df.head()
```

```python
# Lemmatizer
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

def lemmatize_words(text):
    
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])
```

```python
df['reviewText'] = df['reviewText'].apply(lambda x: lemmatize_words(x))
```

```python
df.head()
```

```python
## Train Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['reviewText'], df['rating'], test_size=0.20)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
bow = CountVectorizer()

X_train_bow = bow.fit_transform(X_train).toarray()
X_test_bow = bow.transform(X_test).toarray()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()

X_train_tfidf = tfidf.fit_transform(X_train).toarray()
X_test_tfidf = tfidf.transform(X_test).toarray()
```

```python
X_train_bow
```

```python
from sklearn.naive_bayes import GaussianNB

nb_model_bow = GaussianNB().fit(X_train_bow, y_train)

nb_model_tfidf = GaussianNB().fit(X_train_tfidf, y_train)
```

```python


y_pred_bow = nb_model_bow.predict(X_test_bow)

y_pred_tfidf = nb_model_tfidf.predict(X_test_tfidf)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

y_pred_bow = nb_model_bow.predict(X_test_bow)

# ✅ rename variable to avoid conflict
cm = confusion_matrix(y_test, y_pred_bow)
print("Confusion Matrix:\n", cm)

accuracy = accuracy_score(y_test, y_pred_bow)
print("Accuracy:", accuracy)

print("Classification Report:\n", classification_report(y_test, y_pred_bow))
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

y_pred_tfidf = nb_model_tfidf.predict(X_test_tfidf)

cm_tfidf = confusion_matrix(y_test, y_pred_tfidf)
print("Confusion Matrix:\n", cm_tfidf)

accuracy_tfidf = accuracy_score(y_test, y_pred_tfidf)
print("Accuracy:", accuracy_tfidf)

print("Classification Report:\n", classification_report(y_test, y_pred_tfidf))
```
