```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

data = np.array([['Red'], ['Blue'], ['Green'], ['Blue']])

encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(data)

print(encoded)
print("Feature Names:", encoder.get_feature_names_out())
```

```python
sentence = "I love Data Science"
words = sentence.split()

vocab = sorted(set(words))
print("Vocabulary:", vocab)

one_hot = {word: [1 if w == word else 0 for w in vocab] for word in vocab}

for word, vec in one_hot.items():
    print(f"{word}: {vec}")
```
