### What Are Embeddings?
Think of embeddings as a way to translate words into a language that computers understand - numbers!

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Simplified 2D example (real embeddings have hundreds of dimensions)
word_embeddings = {
    "cat":    [0.82, 0.55],
    "kitten": [0.80, 0.57],
    "dog":    [0.70, 0.40],
    "puppy":  [0.69, 0.42],
    "car":    [-0.50, 0.20],
    "truck":  [-0.52, 0.18]
}
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

for word, coords in word_embeddings.items():
    ax.scatter(coords[0], coords[1], s=100)
    ax.annotate(word, (coords[0], coords[1]), xytext=(5, 5), 
                textcoords='offset points')

ax.set_xlabel('Dimension 1')
ax.set_ylabel('Dimension 2')
ax.set_title('Simplified Word Embeddings in 2D Space')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Measuring Similarity

```python
def cosine_similarity(vec1, vec2):
    """
    Cosine similarity measures the angle between two vectors.
    - Result close to 1: Very similar
    - Result close to 0: Not related
    - Result close to -1: Opposite meanings
    """

    dot_product = np.dot(vec1,vec2)
    norm_a = np.linalg.norm(vec1)
    norm_b = np.linalg.norm(vec2)
    
    return dot_product / (norm_a * norm_b)
```

```python
# Simplified 3D example (real embeddings have hundreds of dimensions)
cat_vector = [0.8, 0.6, 0.3]
kitten_vector = [0.75, 0.65, 0.35]
car_vector = [-0.5, 0.2, 0.1]

cat_kitten_similarity = cosine_similarity(cat_vector, kitten_vector)
# Result close to 1: Very similar
print("Cat and Kitten similarity: ", cat_kitten_similarity)



cat_car_similarity = cosine_similarity(cat_vector, car_vector)
# Result close to -1: Opposite meanings
print("Cat and Car similarity: ", cat_car_similarity)
```

### Creating Your First Embeddings

```python
# Huggingface Embedings

from langchain_huggingface import HuggingFaceEmbeddings

# Huggingface Embeddings initialization
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

print(embeddings)
```

```python
# create your first embeddings
text="Hello, I am learning about embeddings!"

embedding = embeddings.embed_query(text)

print(f"Text: {text} \n")

print(f"Embedding length : {len(embedding)} \n")
print(embedding)
```

```python
sentences = [
    "The cat sat on the mat",
    "The cat sat on the mat",
    "The dog played in the yard",
    "I love programming in Python",
    "Python is my favorite programming language"
]

embedding_sentence = embeddings.embed_documents(sentences)

print(f"Embedding length : {len(embedding_sentence)} \n")

print(embedding_sentence)
print(embedding_sentence[0])
```
