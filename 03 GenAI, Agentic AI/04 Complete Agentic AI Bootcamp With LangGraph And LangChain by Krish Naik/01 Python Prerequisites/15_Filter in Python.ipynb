```python
# The filter() function is used to filter out elements from an iterable based on a condition.
# It only keeps the elements for which the function returns True.
```

```python
def is_even(x):
    return x % 2 == 0

nums = [1, 2, 3, 4, 5, 6]
result = filter(is_even, nums)

print(result)          # <filter object at 0x...>
print(list(result))    # [2, 4, 6]
```

```python
# Using filter() with Lambda 
nums = [10, 15, 20, 25, 30]
result = filter(lambda x: x > 20, nums)
print(list(result))  # [25, 30]
```

```python
# Filtering Strings 
names = ["Alice", "Bob", "Charlie", "David"]

# Only names starting with 'A'
result = filter(lambda name: name.startswith("A"), names)
print(list(result))  # ['Alice']
```

```python
# Removing Empty Strings 
words = ["hello", "", "world", "", "python"]
result = filter(None, words)  # removes empty strings (falsy values)
print(list(result))  # ['hello', 'world', 'python']
```

```python
# Filtering Odd Numbers  
nums = [1, 2, 3, 4, 5, 6, 7, 8, 9]

result = filter(lambda x: x % 2 != 0, nums)
print(list(result))  # [1, 3, 5, 7, 9]
```

```python
# Summary

# filter() → selects elements that satisfy a condition.

# Needs a function that returns True/False.

# Returns an iterator (convert with list() to view).

# Commonly used with lambda functions.
```
