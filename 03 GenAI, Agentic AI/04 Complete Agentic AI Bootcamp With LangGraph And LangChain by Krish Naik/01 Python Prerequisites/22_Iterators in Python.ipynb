```python
# What is an Iterator?

# An iterator is an object that allows you to traverse through all elements of a collection (like list, tuple, etc.) one at a time.

# It implements two special methods:

# __iter__() → Returns the iterator object itself.

# __next__() → Returns the next element. Raises StopIteration when no items are left.
```

```python
# Custom Iterator (Real-World: Countdown)
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self  # returns iterator object itself

    def __next__(self):
        if self.current <= 0:
            raise StopIteration  # stop when finished
        num = self.current
        self.current -= 1
        return num

# Using our iterator
count = Countdown(5)

for num in count:
    print(num)
```
