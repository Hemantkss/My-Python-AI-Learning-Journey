```python
# A list is a collection of items in Python.

# It is ordered, mutable (changeable), allows duplicates, and can hold different data types.
```

```python
# Creating Lists
my_list = [1, 2, 3, 4, 5]
print(my_list)

# Creating a list with different data types
mixed_list = [1, "two", 3.0, True]
print(mixed_list)
```

```python
# Accessing List Elements 
fruits = ["apple", "banana", "cherry", "date"]
print(fruits[0])  # Accessing the first element
print(fruits[1])  # Accessing the second element
print(fruits[-1]) # Accessing the last element
```

```python
# Slicing
print(fruits[0:2])  
print(fruits[1:2])
print(fruits[1:])
print(fruits[:])  
```

```python
# Modifying Lists
fruits = ["apple", "banana", "cherry", "date"]
fruits.append("mango")    # Add at end
fruits.insert(1, "grape") # Insert at index
fruits.remove("banana")   # Remove by value
fruits.pop(0)             # Remove by index
fruits[0] = "kiwi"        # Change element

print(fruits)
```

```python
# List Operations 
a = [1, 2, 3]
b = [4, 5] 

print(a + b)   # Concatenation → [1, 2, 3, 4, 5]
print(a * 2)   # Repetition → [1, 2, 3, 1, 2, 3]
print(3 in a)  # Membership → True
```

```python
# Useful List Methods 
numbers = [3, 1, 4, 1, 5, 9]

print(len(numbers))     # length
print(max(numbers))     # maximum
print(min(numbers))     # minimum
print(numbers.count(1)) # count occurrences
# numbers.sort()          # sort in place
# numbers.reverse()       # reverse in place
print(numbers)

print(numbers.index(3))

```

```python
# List Comprehension in Python
# Short way to create lists.

# Combines for loops and if conditions into one line.
```

```python
squares = [x**2 for x in range(1, 6)]
print(squares)
# ✅ Output: [1, 4, 9, 16, 25]
```

```python
evens = [x for x in range(1, 11) if x % 2 == 0]
print(evens)
# ✅ Output: [2, 4, 6, 8, 10]
```

```python
words = ["apple", "banana", "cherry"]
first_letters = [word[0] for word in words]
print(first_letters)
# ✅ Output: ['a', 'b', 'c']
```

```python
matrix = [[i*j for j in range(1, 4)] for i in range(1, 4)]
print(matrix)
# ✅ Output: [[1, 2, 3], [2, 4, 6], [3, 6, 9]]
```
