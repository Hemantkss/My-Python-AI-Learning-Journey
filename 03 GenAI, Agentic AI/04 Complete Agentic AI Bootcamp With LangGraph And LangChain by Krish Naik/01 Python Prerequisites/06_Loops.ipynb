```python
# A loop allows us to repeat a block of code multiple times.

# Instead of writing the same code again and again, we use loops.

# Python has two main types of loops:

# for loop

# while loop
```

```python
# for Loop 

for i in range(1, 11):
    print(i)
```

```python
for i in range(0, 21, 2):
    print(i)
```

```python
for char in "Python":
    print(char)
```

```python
# while Loop 

i = 1
while i <= 10:
    print(i)
    i += 1
```

```python
n = 5
while n > 0:
    print(n)
    n -= 1
```

```python
# . Loop Control Statements 
```

```python
# break → exit loop immediately

for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
# continue → skip current iteration

for i in range(10):
    if i == 5:
        continue
    print(i)
```

```python
# ✅ (c) pass → placeholder (does nothing) 

for i in range(10):
    if i == 5:
        pass
    print(i)
```

```python
# Nested Loops 

for i in range(3):
    for j in range(2):
        print(f"i: {i}, j: {j}")
```

```python
# Else with Loops 
for i in range(5):
    print(i)
else:
    print("Loop finished!")
```

```python
# calculate sum of first n natural numbers using while and for loop
n = 10
sum= 0
count = 1

while count <= n:
    sum += count
    count += 1
print("Sum using while loop:", sum)

# Resetting variables for for loop
sum = 0
for i in range(1, n + 1):
    sum += i
print("Sum using for loop:", sum)
```

```python
secret = 7
guess = 0

while guess != secret:
    guess = int(input("Guess a number: "))
print("Correct!")
```
