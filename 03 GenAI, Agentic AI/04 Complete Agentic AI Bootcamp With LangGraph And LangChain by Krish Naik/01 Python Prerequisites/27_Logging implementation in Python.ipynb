```python
import logging

# Configure logging
logging.basicConfig(level=logging.DEBUG)

# log MSG
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

```python
# ✅ 2. Logging Levels
# Level	Use Case
# DEBUG	Detailed info (used during development).
# INFO	General info (program is working as expected).
# WARNING	Something unexpected happened (but program continues).
# ERROR	A serious issue, program may not work properly.
# CRITICAL	Very serious error (program may stop).

# 👉 The default level is WARNING (so DEBUG & INFO don’t show unless configured).
```

```python
import logging

# Configure logging 
logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

# log MSG
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

```python
def add(a,b):
    logging.debug(f"Adding {a} and {b}")
    return a + b

logging.debug(f"Result: {add(5,3)}")
```

```python
import logging
# Logging with Multiple Loggers

# create a logger for model1
logger1 = logging.getLogger('model1')
logger1.setLevel(logging.DEBUG)

# create a logger for model2
logger2 = logging.getLogger('model2')
logger2.setLevel(logging.WARNING)

# Configure logging 
logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

# log msg with diffrent loggers
logger1.debug("This is a debug message from model1")
logger2.warning("This is a warning message from model2")
logger2.error("This is an error message from model2")


```
