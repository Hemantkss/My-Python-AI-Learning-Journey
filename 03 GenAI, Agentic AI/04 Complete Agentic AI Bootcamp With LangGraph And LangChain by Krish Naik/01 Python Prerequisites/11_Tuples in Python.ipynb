```python
# A Tuple is a collection that is:

# Ordered (elements keep their position)

# Immutable (cannot be changed after creation)

# Allows duplicate values
```

```python
# Creating Tuples
# Normal tuple
t1 = (10, 20, 30)

# Mixed data types
t2 = ("Rahul", 25, 5.6)

# Nested tuple
t3 = (1, (2, 3), (4, 5))

# Single element tuple → must add a comma
t4 = (5,)  
```

```python
# Accessing Elements 
numbers = (10, 20, 30, 40, 50)

print(numbers[0])   # 10
print(numbers[-1])  # 50 (last element)
print(numbers[1:4]) # (20, 30, 40)
```

```python
# Immutability 
numbers = (10, 20, 30)
# numbers[0] = 100  #❌ Error: 'tuple' object does not support item assignment
# But you can convert tuple → list → modify → tuple again:

temp = list(numbers)
temp[0] = 100
numbers = tuple(temp)
print(numbers)  # (100, 20, 30)
```

```python
# Tuple Operations 
t1 = (1, 2, 3)
t2 = (4, 5)

print(t1 + t2)   # Concatenation → (1, 2, 3, 4, 5)
print(t1 * 2)    # Repetition → (1, 2, 3, 1, 2, 3)
print(2 in t1)   # Membership → True
```

```python
# Tuple Functions 
t = (5, 10, 5, 20)

print(len(t))     # 4 → number of elements
print(max(t))     # 20
print(min(t))     # 5
print(sum(t))     # 40
print(t.count(5)) # 2 → count of element
print(t.index(10))# 1 → position of element
```

```python
# Tuple Unpacking
person = ("Rahul", 25, "Mumbai")

name, age, city = person
print(name)  # Rahul
print(age)   # 25
print(city)  # Mumbai
```

```python
# Nested Tuples 
nested = (1, (2, 3), (4, 5))
print(nested[1][0])  # 2
```

```python
locations = {
    (19.07, 72.87): "Mumbai",
    (28.61, 77.20): "Delhi"
}
print(locations[(19.07, 72.87)])  # Mumbai
```

```python
# Tuples = () (ordered, immutable, allow duplicates).

# Used when data should not change.

# Support indexing, slicing, unpacking, and some built-in functions.
```
