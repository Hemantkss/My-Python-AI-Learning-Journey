```python
# Step 1: Define a custom exception
class InvalidAgeError(Exception):
    """Raised when the age is less than 18"""
    pass

# Step 2: Use it in code
def register_user(age):
    if age < 18:
        raise InvalidAgeError("❌ Age must be at least 18 to register.")
    print("✅ Registration successful!")

# Step 3: Handle exception
try:
    register_user(15)   # Invalid case
except InvalidAgeError as e:
    print("Error:", e)

try:
    register_user(20)   # Valid case
except InvalidAgeError as e:
    print("Error:", e)
```
