```python
# 1. What is a Class?

# A class is a blueprint for creating objects.

# It defines attributes (variables) and methods (functions).

# Think of a class like a design of a car, and objects as actual cars built from that design.

# 2. What is an Object?

# An object is an instance of a class.

# Each object has its own data (attributes) and can perform actions (methods).
```

```python
# Creating a Class
class Car:
    # attribute
    wheels = 4
    doors = 4
    
    # constructor method (special method to initialize)
    def __init__(self, brand, color):
        self.brand = brand    # instance attribute
        self.color = color
    
    # method
    def drive(self):
        return f"{self.brand} car is driving"
```

```python
# Creating Objects 
# Create objects (instances)
car1 = Car("Toyota", "Red")
car2 = Car("BMW", "Black")

print(car1.brand)     # Toyota
print(car2.color)     # Black
print(car1.drive())   # Toyota car is driving


# Difference Between Class Attributes & Instance Attributes
# Class Attribute: Shared by all objects (e.g., wheels).
# Instance Attribute: Unique for each object (brand, color).

print(car1.wheels)  # 4 (shared)
print(car2.wheels)  # 4 (shared)
print(car2.doors)   # 4 (shared)

```

```python
# Special Methods (__init__, __str__, etc.)
# __init__: Constructor (runs when object is created).
# __str__: String representation of object.
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __str__(self):
        return f"Student: {self.name}, Age: {self.age}"

s1 = Student("Alice", 20)
print(s1)   # Calls __str__
```

```python
# Methods in a Class
# Instance Methods → Work with object data (self).
# Class Methods → Work with class data (@classmethod).
# Static Methods → Independent utility functions (@staticmethod).
class Car:
    showroom_name = "SuperCars Showroom"
    total_cars = 0
    
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price
        Car.total_cars += 1
    
    # Instance method
    def car_info(self):
        return f"Brand: {self.brand}, Price: ${self.price}"
    
    # Class method
    @classmethod
    def get_total_cars(cls):
        return f"Total cars in showroom: {cls.total_cars}"
    
    # Static method
    @staticmethod
    def apply_discount(price, discount):
        return price - (price * discount / 100)


# --- Usage ---
car1 = Car("Tesla", 60000)
car2 = Car("BMW", 80000)

# Instance method (specific car info)
print(car1.car_info())   # Brand: Tesla, Price: $60000

# Class method (overall info)
print(Car.get_total_cars())   # Total cars in showroom: 2

# Static method (utility)
print(Car.apply_discount(80000, 15))   # 68000.0

```

# Inheritance in Python (OOP)


```python

# 1. What is Inheritance?

# Inheritance allows one class (child / subclass) to use the properties and methods of another class (parent / superclass).

# It promotes code reusability and extensibility.
```

```python
# Basic Inheritance 
# Parent class
class Vehicle:
    def __init__(self, brand):
        self.brand = brand
    
    def start(self):
        return f"{self.brand} is starting..."
    
# Child class
class Car(Vehicle):
    def drive(self):
        return f"{self.brand} car is driving..."
    

# --- Usage ---
v1 = Vehicle("Generic Vehicle")
print(v1.start())  # Generic Vehicle is starting...

c1 = Car("Tesla")
print(c1.start())  # Tesla is starting... (inherited)
print(c1.drive())  # Tesla car is driving... (own method)
```

```python
# Method Overriding 
# A child class can override (redefine) parent class methods.
class Vehicle:
    def fuel_type(self):
        return "Petrol or Diesel"

class ElectricCar(Vehicle):
    def fuel_type(self):   # overriding
        return "Electric Battery"
    

v = Vehicle()
e = ElectricCar()

print(v.fuel_type())   # Petrol or Diesel
print(e.fuel_type())   # Electric Battery
```

```python
# super() → Access Parent Methods 
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)  # call parent constructor
        self.model = model

    def details(self):
        return f"{self.brand} {self.model}"
    

c = Car("Tesla", "Model S")
print(c.details())   # Tesla Model S
```

```python
# Types of Inheritance in Python
# (a) Single Inheritance  
class A:
    pass
class B(A):
    pass
```

```python
# (b) Multiple Inheritance

# Child inherits from multiple parents. 
class A:
    def featureA(self): return "Feature A"

class B:
    def featureB(self): return "Feature B"

class C(A, B):   # inherits from both A and B
    pass

obj = C()
print(obj.featureA())  # Feature A
print(obj.featureB())  # Feature B
```

```python
# (c) Multilevel Inheritance

# Grandparent → Parent → Child.4
class Grandparent:
    def feature1(self): return "Grandparent feature"

class Parent(Grandparent):
    def feature2(self): return "Parent feature"

class Child(Parent):
    def feature3(self): return "Child feature"

obj = Child()
print(obj.feature1(), obj.feature2(), obj.feature3())
```

```python
# (d) Hierarchical Inheritance

# One parent → many children.  
class Vehicle:
    pass
class Car(Vehicle):
    pass
class Bike(Vehicle):
    pass

    
```

# Polymorphism in Python (OOP)

```python
# Polymorphism = One name, many forms.

# A function/method/operator behaves differently depending on the object or context.

# It allows different classes to define methods with the same name but different implementations.
```

```python
# Polymorphism with Functions and Objects
class Dog:
    def sound(self):
        return "Woof!"

class Cat:
    def sound(self):
        return "Meow!"

# Same function name "sound" works differently for different objects
for animal in (Dog(), Cat()):
    print(animal.sound())
```

```python
# Polymorphism with Inheritance (Method Overriding)
# When a child class overrides a parent method, it creates polymorphism. 
class Vehicle:
    def move(self):
        return "Vehicle is moving"

class Car(Vehicle):
    def move(self):
        return "Car is driving"

class Boat(Vehicle):
    def move(self):
        return "Boat is sailing"

for v in (Vehicle(), Car(), Boat()):
    print(v.move())
```

```python
# Polymorphism with Built-in Functions
# Some built-in functions work on different types (polymorphism):
print(len("Python"))       # 6 (string length)
print(len([1, 2, 3, 4]))   # 4 (list length)
print(len({"a": 1, "b": 2}))  # 2 (dict length)
```

```python
# Polymorphism with Abstract Base Classes (ABC)
# 👉 Best practice: use abstract classes to enforce polymorphism.
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, r): self.r = r
    def area(self): return 3.14 * self.r * self.r

class Square(Shape):
    def __init__(self, s): self.s = s
    def area(self): return self.s * self.s

shapes = [Circle(5), Square(4)]
for shape in shapes:
    print(shape.area())
```

```python
class Payment:
    def pay(self, amount):
        raise NotImplementedError("Subclass must implement this method")

class CreditCard(Payment):
    def pay(self, amount):
        return f"Paid {amount} using Credit Card"

class PayPal(Payment):
    def pay(self, amount):
        return f"Paid {amount} using PayPal"

class Bitcoin(Payment):
    def pay(self, amount):
        return f"Paid {amount} using Bitcoin"

# Polymorphism in action
payments = [CreditCard(), PayPal(), Bitcoin()]

for method in payments:
    print(method.pay(100))
```

# Encapsulation in OOP (Python)

```python
# 👉 Encapsulation = Data (variables) + Methods (functions) bundled inside a single unit (class).

# It restricts direct access to some data.

# Helps achieve data hiding and security.

# Provides controlled access through getters and setters.

# Think of it like a capsule (medicine): ingredients are hidden inside, and you only take it as intended.

# 2. Encapsulation in Python

# Python doesn’t have strict access modifiers like Java/C++ (public, private, protected).
# Instead, it follows conventions:

# Public: Accessible everywhere. (variable)

# Protected: Indicated with _single_underscore. (_variable)

# Private: Indicated with __double_underscore. (__variable)
```

```python
# Example: Public, Protected, Private 
class Employee:
    def __init__(self, name, salary):
        self.name = name          # Public
        self._dept = "IT"         # Protected (convention)
        self.__salary = salary    # Private

    def show(self):
        return f"Name: {self.name}, Dept: {self._dept}, Salary: {self.__salary}"

emp = Employee("Alice", 50000)

print(emp.name)       # ✅ Accessible
print(emp._dept)      # ⚠️ Accessible but should be treated as protected
# print(emp.__salary) # ❌ Error: Private attribute

print(emp.show())     # ✅ Access salary using method
```

```python
# Getter and Setter Methods
# We use getter & setter methods to access or modify private variables.
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # Private

    # Getter
    def get_balance(self):
        return self.__balance

    # Setter
    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Invalid balance!")

acc = BankAccount(1000)
print(acc.get_balance())   # 1000
acc.set_balance(2000)
print(acc.get_balance())   # 2000
acc.set_balance(-500)      # Invalid balance!
```

```python
# Encapsulation with @property (Pythonic way)
# Python provides the @property decorator to make getters/setters more elegant. 

class Student:
    def __init__(self, name, marks):
        self.__marks = marks
        self.name = name

    @property
    def marks(self):
        return self.__marks   # Getter

    @marks.setter
    def marks(self, value):
        if 0 <= value <= 100:
            self.__marks = value
        else:
            print("Invalid marks!")

s = Student("John", 85)
print(s.marks)    # 85
s.marks = 95      # ✅ updates
print(s.marks)    # 95
s.marks = 150     # ❌ Invalid marks!
```

```python
# Real-World Example: ATM Machine
# Encapsulation ensures sensitive data (like PIN, balance) is hidden and accessed only via methods. 

class ATM:
    def __init__(self, pin, balance):
        self.__pin = pin
        self.__balance = balance

    def withdraw(self, pin, amount):
        if pin == self.__pin:
            if amount <= self.__balance:
                self.__balance -= amount
                return f"Withdrawn: {amount}, Balance: {self.__balance}"
            else:
                return "Insufficient balance"
        else:
            return "Invalid PIN"

    def check_balance(self, pin):
        return self.__balance if pin == self.__pin else "Invalid PIN"

atm = ATM("1234", 5000)
print(atm.check_balance("1234"))   # 5000
print(atm.withdraw("1234", 1000))  # Withdrawn: 1000, Balance: 4000
print(atm.check_balance("0000"))   # Invalid PIN
```

# Abstraction in OOPs (Python)

```python
# Abstraction = Hiding implementation details and showing only the necessary features to the user.

# Achieved in Python using the abc module (Abstract Base Class).

# It allows us to define abstract methods (methods without implementation).

# A child class must override these abstract methods.
```

```python
# Example Without Abstraction (Problem) 
class Car:
    def start_engine(self):
        print("Starting engine...")

    def change_gear(self):
        print("Changing gear...")

    def drive(self):
        print("Car is moving")

# Here, every user of Car has to know all the methods (start_engine, change_gear, etc.).
# But what if we want to force every type of Car (e.g., ElectricCar, PetrolCar) to have certain essential methods?
# 👉 That’s where Abstraction helps.
```

```python
# Abstraction with abc Module 
from abc import ABC, abstractmethod

# Abstract Class
class Car(ABC):
    @abstractmethod
    def start(self):
        pass   # Only declared, no implementation

    @abstractmethod
    def stop(self):
        pass

# Concrete Class 1
class ElectricCar(Car):
    def start(self):
        print("Electric Car started with a button 🔘")

    def stop(self):
        print("Electric Car stopped silently 🔋")

# Concrete Class 2
class PetrolCar(Car):
    def start(self):
        print("Petrol Car started with a key 🔑")

    def stop(self):
        print("Petrol Car stopped with engine sound ⛽")

# Using the classes
tesla = ElectricCar()
tesla.start()
tesla.stop()

bmw = PetrolCar()
bmw.start()
bmw.stop()
```

```python
# Why Use Abstraction?

# Forces standardization (all subclasses must implement important methods).
# Hides implementation details → user only calls start() or stop(), doesn’t care how they work internally.
# Increases flexibility (easy to add new car types).
```

```python
# Real-World Example – Payment System 
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} 💳 using Credit Card")

class PayPalPayment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} 🅿️ using PayPal")

# Usage
payment1 = CreditCardPayment()
payment1.pay(500)

payment2 = PayPalPayment()
payment2.pay(1000)
```
