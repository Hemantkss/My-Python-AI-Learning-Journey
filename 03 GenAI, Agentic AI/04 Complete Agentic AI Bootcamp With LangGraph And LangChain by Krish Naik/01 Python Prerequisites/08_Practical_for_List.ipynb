```python
to_do_list = ['Buy groceries', 'Walk the dog', 'Read a book']

# Add a new task
to_do_list.append('Exercise for 30 minutes')
to_do_list.append('Call mom')

# remove a task
to_do_list.remove('Read a book')

# checking a task in the list
if 'Walk the dog' in to_do_list:
    print('Task found: Walk the dog')
    
print('Current To-Do List:')

for task in to_do_list:
    print('-', task)
```

```python
# Organizing student grades 

grades = [85, 90, 78, 92, 88]

# Adding a new grade
grades.append(95)

# Calculating avg of all grades
average = sum(grades) / len(grades)
print(f'Average Grade: {average}')

# Finding the highest and lowest grades
highest = max(grades)
lowest = min(grades)
print(f'Highest Grade: {highest}')
print(f'Lowest Grade: {lowest}')
```

```python
# manage inventory 
inventory = ['apple', 'banana', 'orange']

# Adding a new item
inventory.append('grape')

# Removing an item
inventory.remove('banana')

# Checking if an item exists
item = 'apple'
if item in inventory:
    print(f'{item} is in the inventory.')
else:
    print(f'{item} is not in the inventory.')
    
print('Current Inventory:')
for item in inventory:
    print('-', item)
```

```python
# Collecting users feedback 
feedback = ["Great service!", "Very satisfied with the product.", "Could be improved."]

# Adding new feedback
feedback.append("Loved the user interface!")

# Counting specific feedback
positive_feedback = sum(1 for comment in feedback if "great" in comment.lower() or "satisfied" in comment.lower())
print(f'Number of positive feedbacks: {positive_feedback}')

# print all feedback
print('All Feedback:')
for comment in feedback:
    print('-', comment)
```
