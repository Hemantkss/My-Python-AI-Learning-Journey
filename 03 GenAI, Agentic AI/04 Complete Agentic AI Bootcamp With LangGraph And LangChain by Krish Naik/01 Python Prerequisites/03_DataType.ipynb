```python
# Basic Data Types in Python

# Integer
age = 25
print(type(age))

# Float
height = 5.9
print(type(height))

# String
name = "John"
print(type(name))

# Boolean
is_student = True
print(type(is_student))

# Complex
complex_num = 1 + 2j
print(type(complex_num))
```

```python
# Collection and sequence data types

# List
fruits = ["apple", "banana", "cherry"]
print(type(fruits))

# Tuple
coordinates = (10.0, 20.0)
print(type(coordinates))

# Set
unique_numbers = {1, 2, 3, 4, 5}
print(type(unique_numbers))

# Dictionary
person = {"name": "Alice", "age": 30}
print(type(person))
```

```python
# Special Types

# NoneType
nothing = None
print(type(nothing))

# Ellipsis
something = ...
print(type(something))

# NotImplemented
not_implemented = NotImplemented
print(type(not_implemented))
```
