## OpenAI Embeddings

```python
import os 
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(model='text-embedding-3-large')
# embeddings = OpenAIEmbeddings(model='text-embedding-3-large', dimensions=1024)

```

```python
text = "I love programming in Python"
query_result = embeddings.embed_query(text)
query_result
```

```python
print(len(query_result))  # Length of the embedding vector
print(query_result[:5])  # Print the first 5 dimensions of the embedding vector
print(type(query_result))  # Type of the embedding vector
```

```python
# Text File Embedding
from langchain_community.document_loaders import TextLoader

loader = TextLoader('speech.txt')
docs = loader.load()
docs
```

```python
# Text Splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=20)
final_doc = text_splitter.split_documents(docs)
final_doc
```

```python
# Vector Embedding And Vectore Store
from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(final_doc, embedding=embeddings)
db
```

```python
# Similarity Search
query = "It will be all the easier for us to conduct ourselves as belligerents in a high"

docs = db.similarity_search(query)
print(docs)
```
