#### Chroma
Chroma is a AI-native open-source vector database focused on developer productivity and happiness. Chroma is licensed under Apache 2.0.

https://python.langchain.com/v0.2/docs/integrations/vectorstores/

```python
from langchain_community.document_loaders import TextLoader
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
# Text loader
loader = TextLoader('speech.txt')
docs = loader.load()

# Text splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=30
)

final_split = text_splitter.split_documents(docs)
final_split
```

```python
# Embedding 
embeddings = OpenAIEmbeddings(
    model='text-embedding-3-large',
    dimensions=1024
)

# Vector DB
db = Chroma.from_documents(
    documents=final_split,
    embedding=embeddings
)
db
```

```python
# Query 
query = "What is Agentic AI?"

docs = db.similarity_search(query, k=2)
print(docs[0].page_content)
```

```python
# Saving the Vector Database in Local
db = Chroma.from_documents(
    documents=final_split,
    embedding=embeddings,
    persist_directory="./Chroma_db"
)


# Loading the Vector Database from Local
new_db = Chroma(persist_directory="./Chroma_db", embedding_function=embeddings)
```

```python
# similarity search
query = "What is  Langsmith?"
docs = new_db.similarity_search(query, k=2)
print(docs[0].page_content)
```

```python
# silimarity search with Score
query = "What is  Langsmith?"
docs = new_db.similarity_search_with_score(query, k=2)
print(docs[0])
```

```python
# search in new_db using Retriever
query = "What is  Langsmith?"

# search using Retriever
retriever=new_db.as_retriever()
retriever.invoke(query)[0].page_content
```
