#### Faiss
Facebook AI Similarity Search (Faiss) is a library for efficient similarity search and clustering of dense vectors. It contains algorithms that search in sets of vectors of any size, up to ones that possibly do not fit in RAM. It also contains supporting code for evaluation and parameter tuning.

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Text Loader
loader = TextLoader('speech.txt')
docs = loader.load()

# Text Splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=30
)

final_split = text_splitter.split_documents(docs)
```

```python
final_split[:5]
```

```python
# Embeddings
embeddings = OpenAIEmbeddings(model='text-embedding-3-large', dimensions=1024)

# Vector Database
vector_db = FAISS.from_documents(final_split, embeddings)
vector_db
```

```python
# Query
query = "What is LangGraph?"

# Similarity Search
similarity_search = vector_db.similarity_search(query)
print(similarity_search[0].page_content)
```

```python
# Query
query = "What is LangGraph?"

# similarity search with Score
similarity_search_with_score = vector_db.similarity_search_with_score(query)
print(similarity_search_with_score[0])
```

```python
# Example with using Retriever
# We can also convert the vectorstore into a Retriever class. This allows us to easily use it in other LangChain methods, which largely work with retriever
```

```python
query = "What is LangChain?"
retriever = vector_db.as_retriever()
final_output = retriever.invoke(query)
print(final_output[0].page_content)
```

```python
# query
query = "What is LangChain?"

# Embeddings
embedding_vector=embeddings.embed_query(query)

docs_score=vector_db.similarity_search_by_vector(embedding_vector)
print(docs_score[0])
```

```python
# Saving the Vector Database in Local

vector_db.save_local("FAISS_db")

# Loading the Vector Database from Local
new_vector_db = FAISS.load_local("FAISS_db", embeddings, allow_dangerous_deserialization=True)

# Search in new Vector Database using similarity search
query = "What is LangGraph?"
similarity_search = new_vector_db.similarity_search(query)
print(similarity_search[0].page_content)
```
