```python
# Basic Example 
from pydantic import BaseModel

class User(BaseModel):
    id: int
    name: str
    age: int
    is_active: bool = True   # default value

# ✅ Valid Data
user = User(id="1", name="Hemant", age="23")
print(user)  
print(user.id)   # Automatically converted str -> int

# ❌ Invalid Data
try:
    User(id="abc", name="Hemant", age="23")
except Exception as e:
    print(e)
    
# id="1" (string) → auto converted to integer.
# Wrong data (id="abc") → Pydantic throws validation error.
```

```python
# Field Validation (You can add constraints to fields)
from pydantic import BaseModel, Field

class Product(BaseModel):
    name: str = Field(..., min_length=3, max_length=50)
    price: float = Field(..., gt=0)  # must be > 0
    quantity: int = Field(..., ge=1) # must be >= 1

p = Product(name="Laptop", price=599.99, quantity=10)
print(p)

# ❌ Wrong Example
# Product(name="L", price=-10, quantity=0)
```

```python
# Custom Validation (You can add custom logic)
from pydantic import BaseModel, validator

class User(BaseModel):
    name: str
    age: int

    @validator("age")
    def check_age(cls, v):
        if v < 18:
            raise ValueError("Age must be 18 or older")
        return v

User(name="Hemant", age=20)   # ✅
# User(name="Kid", age=15)      # ❌ raises error
```

```python
# Nested Models (Pydantic supports nested structures (great for APIs/JSON))
from pydantic import BaseModel
from typing import List

class Address(BaseModel):
    city: str
    zipcode: str

class User(BaseModel):
    name: str
    addresses: List[Address]

data = {
    "name": "Hemant",
    "addresses": [
        {"city": "Pune", "zipcode": "411001"},
        {"city": "Mumbai", "zipcode": "400001"}
    ]
}

user = User(**data)
print(user)

```
